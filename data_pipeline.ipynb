```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

# Define an empty DataFrame
df = pd.DataFrame(columns=[
    "title",
    "price",
    "star_rating",
    "availability",
    "category"
])

# Website URL
base_url = "https://books.toscrape.com/catalogue/"

# List to store scraped book data
books = []

# Scrape first 5 pages
for page in range(1, 6):

    # Page 1 has a different URL
    if page == 1:
        url = base_url + "page-1.html"
    else:
        url = base_url + f"page-{page}.html"

    # Send request
    response = requests.get(url)

    # Check whether request was successful
    response.raise_for_status()

    # Parse HTML
    soup = BeautifulSoup(response.text, "html.parser")

    # Find all books on the page
    products = soup.find_all("article", class_="product_pod")

    # Extract information from each book
    for product in products:

        # Title
        title = product.find("h3").find("a")["title"]

        # Price
        price = product.find("p", class_="price_color").text.strip()

        # Star rating
        rating_class = product.find("p", class_="star-rating")["class"]
        star_rating = rating_class[1]

        # Availability
        availability = product.find(
            "p", class_="instock availability"
        ).get_text(strip=True)

        # Category
        # Category is obtained from the individual book page
        book_link = product.find("h3").find("a")["href"]

        book_url = requests.compat.urljoin(url, book_link)

        book_response = requests.get(book_url)
        book_response.raise_for_status()

        book_soup = BeautifulSoup(
            book_response.text,
            "html.parser"
        )

        breadcrumb = book_soup.find("ul", class_="breadcrumb")

        category = breadcrumb.find_all("li")[-2].get_text(strip=True)

        # Store the book information
        books.append({
            "title": title,
            "price": price,
            "star_rating": star_rating,
            "availability": availability,
            "category": category
        })


# Convert scraped data into DataFrame
df = pd.DataFrame(books)

# Display the first 10 rows
print(df.head(10))
```

```python
import pandas as pd
import sqlite3

# ============================================
# 1. LOAD DATA FROM TASK 1
# ============================================

# If df already exists from Task 1, use it directly.
# Otherwise:
# df = pd.read_csv("books_data.csv")

clean_df = df.copy()


# ============================================
# 2. CLEAN PRICE
# ============================================

clean_df["price_gbp"] = (
    clean_df["price"]
    .astype(str)
    .str.replace("£", "", regex=False)
    .str.strip()
)

clean_df["price_gbp"] = pd.to_numeric(
    clean_df["price_gbp"],
    errors="coerce"
)


# ============================================
# 3. CLEAN RATING
# ============================================

rating_map = {
    "One": 1,
    "Two": 2,
    "Three": 3,
    "Four": 4,
    "Five": 5
}

clean_df["rating"] = clean_df["star_rating"].map(rating_map)


# ============================================
# 4. CLEAN AVAILABILITY
# ============================================

clean_df["in_stock"] = (
    clean_df["availability"]
    .astype(str)
    .str.strip()
    .str.lower()
    .str.contains("in stock", na=False)
)


# ============================================
# 5. MEDIAN IMPUTATION
# ============================================

clean_df["price_gbp"] = clean_df["price_gbp"].fillna(
    clean_df["price_gbp"].median()
)

clean_df["rating"] = clean_df["rating"].fillna(
    clean_df["rating"].median()
)

clean_df["rating"] = clean_df["rating"].round().astype(int)


# ============================================
# 6. GBP → INR
# ============================================

GBP_TO_INR = 105.50

clean_df["price_inr"] = (
    clean_df["price_gbp"] * GBP_TO_INR
).round(2)


# ============================================
# 7. SELECT FINAL COLUMNS
# ============================================

clean_df = clean_df[[
    "title",
    "price_gbp",
    "price_inr",
    "rating",
    "in_stock",
    "category"
]]

print(clean_df.head())


# ============================================
# 8. CREATE SQLITE DATABASE
# ============================================

conn = sqlite3.connect("books.db")

cursor = conn.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS categories (
    category_id INTEGER PRIMARY KEY,
    category_name TEXT UNIQUE NOT NULL
)
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS books (
    book_id INTEGER PRIMARY KEY AUTOINCREMENT,
    title TEXT NOT NULL,
    price_gbp REAL,
    price_inr REAL,
    rating INTEGER,
    in_stock INTEGER,
    category_id INTEGER,
    FOREIGN KEY (category_id)
        REFERENCES categories(category_id)
)
""")

conn.commit()


# ============================================
# 9. INSERT CATEGORIES
# ============================================

for category in clean_df["category"].dropna().unique():

    cursor.execute(
        """
        INSERT OR IGNORE INTO categories (category_name)
        VALUES (?)
        """,
        (category,)
    )

conn.commit()


# ============================================
# 10. INSERT BOOKS
# ============================================

for _, row in clean_df.iterrows():

    cursor.execute(
        """
        SELECT category_id
        FROM categories
        WHERE category_name = ?
        """,
        (row["category"],)
    )

    category_id = cursor.fetchone()[0]

    cursor.execute(
        """
        INSERT INTO books
        (title, price_gbp, price_inr, rating, in_stock, category_id)
        VALUES (?, ?, ?, ?, ?, ?)
        """,
        (
            row["title"],
            row["price_gbp"],
            row["price_inr"],
            row["rating"],
            int(row["in_stock"]),
            category_id
        )
    )

conn.commit()


# ============================================
# 11. SQL QUERY 1: SELECT / WHERE
# ============================================

query1 = """
SELECT title, price_gbp, rating
FROM books
WHERE rating = 5
"""

result1 = pd.read_sql(query1, conn)

print("\nQUERY 1")
display(result1)


# ============================================
# 12. SQL QUERY 2: ORDER BY
# ============================================

query2 = """
SELECT title, price_gbp, rating
FROM books
ORDER BY price_gbp DESC
"""

result2 = pd.read_sql(query2, conn)

print("\nQUERY 2")
display(result2)


# ============================================
# 13. SQL QUERY 3: LIMIT
# ============================================

query3 = """
SELECT title, price_gbp
FROM books
ORDER BY price_gbp DESC
LIMIT 10
"""

result3 = pd.read_sql(query3, conn)

print("\nQUERY 3")
display(result3)


# ============================================
# 14. SQL QUERY 4: DISTINCT
# ============================================

query4 = """
SELECT DISTINCT category_name
FROM categories
ORDER BY category_name
"""

result4 = pd.read_sql(query4, conn)

print("\nQUERY 4")
display(result4)


# ============================================
# 15. SQL QUERY 5: BETWEEN
# ============================================

query5 = """
SELECT title, price_gbp, rating
FROM books
WHERE price_gbp BETWEEN 20 AND 40
ORDER BY price_gbp
"""

result5 = pd.read_sql(query5, conn)

print("\nQUERY 5")
display(result5)


# ============================================
# 16. SQL QUERY 6: JOIN
# ============================================

join_query = """
SELECT
    b.title,
    c.category_name,
    b.price_gbp,
    b.rating,
    b.in_stock
FROM books b
JOIN categories c
    ON b.category_id = c.category_id
ORDER BY b.rating DESC, b.title
LIMIT 10
"""

join_result = pd.read_sql(join_query, conn)

print("\nQUERY 6 - JOIN")
display(join_result)


# ============================================
# 17. REPRODUCE JOIN USING pd.merge()
# ============================================

books_df = pd.read_sql(
    "SELECT * FROM books",
    conn
)

categories_df = pd.read_sql(
    "SELECT * FROM categories",
    conn
)

merge_result = pd.merge(
    books_df,
    categories_df,
    on="category_id",
    how="inner"
)

merge_result = merge_result[[
    "title",
    "category_name",
    "price_gbp",
    "rating",
    "in_stock"
]]

merge_result = merge_result.sort_values(
    by=["rating", "title"],
    ascending=[False, True]
).head(10)

merge_result = merge_result.reset_index(drop=True)


# ============================================
# 18. COMPARE SQL JOIN AND PANDAS MERGE
# ============================================

sql_compare = join_result.reset_index(drop=True)

print("\nSQL JOIN RESULT")
display(sql_compare)

print("\nPANDAS MERGE RESULT")
display(merge_result)

print(
    "\nAre SQL JOIN and pandas merge equivalent?",
    sql_compare.equals(merge_result)
)


# ============================================
# 19. SAVE QUERY OUTPUTS
# ============================================

query_outputs = {
    "query1_select_where": result1,
    "query2_order_by": result2,
    "query3_limit": result3,
    "query4_distinct": result4,
    "query5_between": result5,
    "query6_join": join_result
}

for name, result in query_outputs.items():
    result.to_csv(f"{name}.csv", index=False)


# ============================================
# 20. FINAL VALIDATION
# ============================================

print("\n===== FINAL VALIDATION =====")

print("Total books:", len(clean_df))

print(
    "Books in SQLite:",
    pd.read_sql(
        "SELECT COUNT(*) AS count FROM books",
        conn
    )["count"][0]
)

print(
    "Categories in SQLite:",
    pd.read_sql(
        "SELECT COUNT(*) AS count FROM categories",
        conn
    )["count"][0]
)

print("\nMissing values:")
print(clean_df.isnull().sum())

print("\nRating values:")
print(sorted(clean_df["rating"].unique()))

print(
    "\nJOIN equivalence:",
    sql_compare.equals(merge_result)
)

# Close database
conn.close()
```
